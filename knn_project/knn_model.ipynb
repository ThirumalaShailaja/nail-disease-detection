```python
import os
import cv2
import numpy as np

train_path = r"C:\Users\PC\Desktop\final_project\train"
img_size = 64   # small size for KNN

X = []
y = []

for label in os.listdir(train_path):
    folder_path = os.path.join(train_path, label)
    
    if not os.path.isdir(folder_path):
        continue
        
    for img_name in os.listdir(folder_path):
        img_path = os.path.join(folder_path, img_name)
        
        img = cv2.imread(img_path)
        if img is None:
            continue
            
        img = cv2.resize(img, (img_size, img_size))
        img = img.flatten()
        
        X.append(img)
        y.append(label)

X = np.array(X)
y = np.array(y)

print("✅ Total images loaded:", len(X))
print("✅ Feature shape:", X.shape)
print("✅ Total classes:", len(set(y)))
```

```python
def load_images(folder_path, img_size=64):
    X = []
    y = []
    
    for label in os.listdir(folder_path):
        class_path = os.path.join(folder_path, label)
        
        if not os.path.isdir(class_path):
            continue
            
        for img_name in os.listdir(class_path):
            img_path = os.path.join(class_path, img_name)
            
            img = cv2.imread(img_path)
            if img is None:
                continue
                
            img = cv2.resize(img, (img_size, img_size))
            img = img.flatten()
            
            X.append(img)
            y.append(label)
            
    return np.array(X), np.array(y)

# Load train data
X_train, y_train = load_images("train")

# Load validation data
X_val, y_val = load_images("validation")

print("✅ Train samples:", len(X_train))
print("✅ Validation samples:", len(X_val))
```

```python
# Paste your real paths here
train_path = r"C:\Users\PC\Desktop\final_project\train"
val_path   = r"C:\Users\PC\Desktop\final_project\validation"

X_train, y_train = load_images(train_path)
X_val, y_val     = load_images(val_path)

print("✅ Train samples:", len(X_train))
print("✅ Validation samples:", len(X_val))
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# create KNN model
knn = KNeighborsClassifier(n_neighbors=3)

# train model
knn.fit(X_train, y_train)

print("✅ Model training completed")

# predict on validation data
y_pred = knn.predict(X_val)

# accuracy
acc = accuracy_score(y_val, y_pred)
print("🎯 Validation Accuracy:", acc)
```

```python
import os
import cv2
import numpy as np
from skimage.feature import hog
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Function to load images and extract HOG features
def load_hog_images(folder_path, img_size=64):
    X = []
    y = []
    for label in os.listdir(folder_path):
        class_path = os.path.join(folder_path, label)
        if not os.path.isdir(class_path):
            continue
        for img_name in os.listdir(class_path):
            img_path = os.path.join(class_path, img_name)
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)  # HOG works on grayscale
            if img is None:
                continue
            img = cv2.resize(img, (img_size, img_size))
            features, _ = hog(img, orientations=9, pixels_per_cell=(8,8),
                              cells_per_block=(2,2), visualize=True)
            X.append(features)
            y.append(label)
    return np.array(X), np.array(y)

# Load train and validation with HOG
X_train, y_train = load_hog_images(r"C:\Users\PC\Desktop\final_project\train"
)
X_val, y_val = load_hog_images(r"C:\Users\PC\Desktop\final_project\validation")

print("✅ Train samples:", len(X_train), "Feature length:", X_train.shape[1])
print("✅ Validation samples:", len(X_val), "Feature length:", X_val.shape[1])

# Train KNN with distance weighting
knn = KNeighborsClassifier(n_neighbors=7, weights='distance')
knn.fit(X_train, y_train)

# Predict on validation
y_pred = knn.predict(X_val)

# Accuracy
acc = accuracy_score(y_val, y_pred)
print("🎯 HOG + KNN Validation Accuracy:", acc)
```
